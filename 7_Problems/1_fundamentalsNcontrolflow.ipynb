# Q1: Student Profile Analyzer

Create a program that stores a student's name, age, course, and marks in different variables.

The program should:

- Display the student's information using an f-string.
- Calculate the student's average marks.
- Determine whether the student has passed.
- Classify the student's performance based on the average.

---

```python
name=None
age=None
course=None
marks1=None
marks2=None
marks3=None
has_passed=False

name=input("Enter your name: ")
age=int(input("Enter your age: "))
course=input("Enter your course: ")
marks1=int(input("Enter your marks in subject 1 out of 100: "))
marks2=int(input("Enter your marks in subject 2 out of 100: "))
marks3=int(input("Enter your marks in subject 3 out of 100: "))

total_marks=marks1+marks2+marks3
percentage=(total_marks/300)*100
avg_marks=total_marks/3

if percentage>=40 and marks1>=33 and marks2>=33 and marks3>=33:
    has_passed=True


if avg_marks >= 90:
    performance = "Excellent"
elif avg_marks >= 75:
    performance = "Very Good"
elif avg_marks >= 50:
    performance = "Good"
elif avg_marks >= 40:
    performance = "Average"
else:
    performance = "Needs Improvement"

print("----------")
print(f"Student Profile Summary \nName: {name}, Age: {age}, Course: {course} \nTotal Marks Obtained: {total_marks} \tPercentage: {percentage:.2f}% \nAverage Marks: {avg_marks:.2f} \tPerformance: {performance} \nHas Passed: {'Yes' if has_passed else 'No'}") 
#.2f rounds the decimal to 2 places
print("----------")
```

# Q2: Smart Calculator

Create a calculator that accepts two numbers and an operator (+, -, *, /, %, **).

The program should:
- Perform the selected operation.
- Handle division by zero.
- Display an appropriate message for an invalid operator.

---

```python
x=int(input("Enter First Number"))
y=int(input("Enter Second Number"))

opt=input("Select any one of the operation +, -, *, /, %, **")

if opt == "+" :
    print(f"The addition of {x} and {y} is: {x + y}")
elif opt == "-":
    print(f"The subtraction of {x} and {y} is: {x - y}")
elif opt == "*":
    print(f"The multiplication of {x} and {y} is: {x * y}")
elif opt == "/":
    if y == 0:
        print(f"Denominator should not be Zero!")
    else:
        print(f"The Division of {x} and {y} is: {x / y}")
elif opt == "%":
    if y != 0:
        print(f"The Remainder of {x} by {y} is: {x % y}")
    else:
        print(f"The Denominator should not be Zero!")
elif opt == "**":
    print(f"{x} to the power {y} is: {x ** y}")
else:
    print(f"Invaild! Choose from these operations +, -, *, /, %, ** to perform.")
```

```python
# Right now, the code handles division by zero perfectly. But look at the power operator (**) block
# If a user enters x = 0 and y = -2 (a negative number), Python will try to calculate 0 ** -2.
# In math, a negative exponent means "1 divided by that number". So, 0^(-2) is the exact same thing as 1/0^2, which is 1/0.
# This will result in a ZeroDivisionError. As 0 cannot be raised to power of negative number.
# The fix: 

if x == 0 and y < 0:
        print("Zero cannot be raised to a negative power!")
else:
    print(f"{x} to the power {y} is: {x ** y}")
```

# Q3: Number Classifier

Given an integer, determine whether it is:

- Positive, negative, or zero.
- Even or odd.
- Divisible by 3.
- Divisible by 5.
- Divisible by both 3 and 5.

---

```python
num = int(input("Enter a number"))

if num > 0 or num == 0:
    print("The number is positive.")
    if num == 0:
        print(f"The number {num} is neither even nor odd.")
    elif num % 2 == 0:
        print(f"The number {num} is even.")
    else:
        print(f"The number {num} is odd.")

    if num % 3 == 0 and num % 5 == 0:
        print(f"The number {num} is divisible by both 3 and 5.")
    elif num % 3 == 0:
        print(f"The number {num} is divisible by 3.")
    elif num % 5 == 0:
        print(f"The number {num} is divisible by 5.")
    else:
        print(f"The number {num} is not divisible by either 3 or 5.")
else:
    print("The number is negative.")

```

# Q5: Password Strength Checker

Given a password, determine whether it satisfies the following:

- At least 8 characters.
- Contains an uppercase letter.
- Contains a lowercase letter.
- Contains a digit.

Display which requirements are missing.

---

```python
strpassword = input("Enter a password")

has_upper = False
has_lower = False
has_digit = False

if len(strpassword) >= 8:
    for char in strpassword:
        if char.isupper():
            has_upper = True
        if char.islower():
            has_lower = True
        if char.isdigit():
            has_digit = True

    if has_upper and has_lower and has_digit:
        print("The password is strong.")
    else:
        print("The password is weak. It should contain at least 8 characters and at least one uppercase letter, one lowercase letter, and one digit.")
else:
    print("The password is weak. It should contain at least 8 characters and at least one uppercase letter, one lowercase letter, and one digit.")
```

```python
# Other way to check is using any().
# It returns True if at least one item ina list matched your rule.

text = "Password#1"
has_upper = any(char.isupper() for char in text)
has_lower = any(char.islower() for char in text)
has_digit = any(char.isdigit() for char in text)

print(f"Has Upper, Lower, Digit = {has_upper}, {has_lower}, {has_digit}")
```

# Q6: Number Range Analyzer

Given a number n, iterate from 1 to n and calculate:

- Sum of all numbers.
- Sum of even numbers.
- Sum of odd numbers.
- Count of numbers divisible by 3.
- Count of numbers divisible by 5.

---

```python
num = int(input("Enter a number"))
print(f"You entered: {num}")

sum = 0
for i in range(1, num+1):
    sum +=i

print(f"The sum of all numbers from 1 to {num} is: {sum}")

even_sum=0
odd_sum=0
for i in range(1, num+1):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print(f"The sum of all even numbers from 1 to {num} is: {even_sum}")
print(f"The sum of all odd numbers from 1 to {num} is: {odd_sum}")

count_three = 0
count_five = 0

for i in range(1, num+1):
    if i % 3 == 0:
        count_three += 1
    if i % 5 == 0:
        count_five += 1

print(f"The count of numbers divisible by 3 from 1 to {num} is: {count_three}")
print(f"The count of numbers divisible by 5 from 1 to {num} is: {count_five}")
```

# Q7: Mini Challenge: ATM Transaction

Create a simple ATM program.

The user starts with a balance and chooses:

- Check balance
- Deposit
- Withdraw

The program should validate the transaction and prevent withdrawal when the balance is insufficient.

---

```python
amount= float(input("Enter the amount: "))

if amount < 0:
    print("Amount cannot be negative.")
else:
    while(True):
        print("Select what you want to do with the amount!")
        print("1. Add money")
        print("2. Withdraw money")
        print("3. Check balance")
        print("4. Exit")
        choice = int(input("Enter your choice: "))

        if choice == 1:
            add_amount = float(input("Enter the amount to add: "))
            if add_amount < 0:
                print("Amount cannot be negative.")
            else:
                amount += add_amount
                print(f"Amount added successfully. New balance is: {amount}")

        if choice == 2:
            withdraw_amount = float(input("Enter the amount to withdraw: "))
            if withdraw_amount < 0:
                print("Amount cannot be negative.")
            elif withdraw_amount > amount:
                print("Insufficient balance.")
            else:
                amount -= withdraw_amount
                print(f"Amount withdrawn successfully. New balance is: {amount}")

        if choice == 3:
            print(f"Your current balance is: {amount}")

        if choice == 4:
            print("Thank you for using our service!")
            break
```
