# Q1: Student Data Persistence

Build a small student system that stores student information in a file.

Your program should provide functions such as:

- add_student()
- save_students()
- load_students()
- search_student()
- display_students()

The important behavior is:

Program starts

      ↓

Load existing students

      ↓

Add / search / display

      ↓

Save changes

      ↓

Program closes

Initial Student Data:
```python

students = [
    {
        "id": 101,
        "name": "Arsh",
        "marks": [85, 90, 88]
    },
    {
        "id": 102,
        "name": "Rahul",
        "marks": [72, 75, 81]
    },
    {
        "id": 103,
        "name": "Aman",
        "marks": [91, 95, 89]
    }
]
```

Test operations:-

Add:
- ID = 104
- Name = Neha
- Marks = 82, 87, 90

Search:
- ID = 102

Display:
- All students

Save the data so that it is still present after restarting the program.

---

````python
import json
import os

def load_students():
    if not os.path.exists("f1.json"):
        return []
    
    with open("f1.json", "r") as file:
        return json.load(file)


def save_students(students):
    with open("f1.json", "w") as file:
        json.dump(students, file, indent=4)


def add_student():
    data = input("Enter ID, Marks (space separated), and Name:\nExample: 104 88 90 85 Sneha\n→ ").split()

    student_id = int(data[0])
    *marks, name = data[1:]          # last item is name, rest are marks
    marks = list(map(int, marks))    # convert marks to integers

    new_student = {
        "id": student_id,
        "name": name,
        "marks": marks
    }

    students = load_students()       # Load existing students
    students.append(new_student)     # Add new one
    save_students(students)          # Save back to file

    print("Student added successfully!")


def display_students():
    students = load_students()
    
    if not students:
        print("No students found.")
        return
    
    for student in students:
        print(f"ID: {student['id']} | Name: {student['name']} | Marks: {student['marks']}")


def search_student():
    students = load_students()
    
    if not students:
            print("No students found.")
            return

    searchId = int(input("Enter the ID you want to search: "))

    for student in students:
        if student['id'] == searchId :
            print(f"ID: {student['id']} Found!")
            print(f"Name: {student['name']} | Marks: {student['marks']}")
            return 

    print(f"No student found with ID {searchId}")   

if __name__ == "__main__":
    add_student()
    print("----------")
    display_students()
    print("----------")
    search_student()
````

# Q2: CSV Student Analyzer

Create a file named:

students.csv

with the following data:
```
name,python,sql,dsa
Arsh,85,78,92
Rahul,72,88,75
Aman,95,91,89
Vijay,60,68,64
Neha,82,79,85
```

Create functions:

- read_students()
- calculate_student_average()
- find_top_student()
- find_subject_topper()
- find_passing_students()
- generate_report()

Requirements

Use the CSV as the source of data.

The program should be able to determine:

- Each student's average
- Highest overall student
- Python topper
- SQL topper
- DSA topper
- Students who passed
- Overall class average

Do not hard-code the results.

---

### How Python Reads CSV

We use Python’s built-in csv module.

> Most useful tool: csv.DictReader

```Python
import csv

with open("students.csv", "r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        print(row)
Output:
Python{'name': 'Arsh', 'python': '85', 'sql': '78', 'dsa': '92'}
{'name': 'Rahul', 'python': '72', 'sql': '88', 'dsa': '75'}
...
```

Important points:

- Each row is a dictionary
- Keys come from the header row (name, python, sql, dsa)
- All values are strings by default (even the numbers)

---

````python
import csv

def read_students(filename="students.csv"):
    students = []
    with open(filename, "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            students.append({
                "name": row["name"],
                "python": int(row["python"]),
                "sql": int(row["sql"]),
                "dsa": int(row["dsa"])
            })
    return students


def calculate_student_average(student):
    total = student["python"] + student["sql"] + student["dsa"]
    average = total / 3
    return round(average, 2)


def find_top_student(students):
    top_student = max(students, key=calculate_student_average)
    return top_student


def find_subject_topper(students, subject):
    topper = max(students, key=lambda s: s[subject])
    return topper


def find_passing_students(students, passing_marks=70):
    passing = []
    for student in students:
        if calculate_student_average(student) >= passing_marks:
            passing.append(student["name"])
    return passing


def generate_report(students):
    print("=" * 50)
    print("STUDENT REPORT")
    print("=" * 50)

    # Each student's average
    print("\n--- Student Averages ---")
    for student in students:
        avg = calculate_student_average(student)
        print(f"{student['name']}: {avg}")

    # Top student
    top = find_top_student(students)
    print(f"\nHighest Overall Student: {top['name']} ({calculate_student_average(top)})")

    # Subject toppers
    print("\n--- Subject Toppers ---")
    for subject in ["python", "sql", "dsa"]:
        topper = find_subject_topper(students, subject)
        print(f"{subject.upper()} Topper: {topper['name']} ({topper[subject]})")

    # Passing students
    passing = find_passing_students(students)
    print(f"\nStudents who passed: {', '.join(passing)}")

    # Overall class average
    total_avg = sum(calculate_student_average(s) for s in students) / len(students)
    print(f"\nOverall Class Average: {round(total_avg, 2)}")
    print("=" * 50)


if __name__ == "__main__":
    students = read_students("students.csv")
    generate_report(students)
````

# Q3: Robust Calculator With Exception Handling

Create a calculator using separate functions.

Support:
```
+
-
*
/
%
**
```

Handle:

- Division by zero
- Invalid operator
- Empty input

Use:

- except

- finally

appropriately.

```

```

```

```

```
Test Input 4

```

```
Test Input 5

```

Do not allow the application to terminate unexpectedly because of bad input.

---

````python
def get_number(prompt):
    """
    - Uses a `while True` loop to keep asking until the input is correct.
    - Checks for empty input.
    - Tries to convert the input into a `float`.
    - If conversion fails, it catches `ValueError` and asks again.
    """
    while True:
        try:
            value = input(prompt).strip()
            
            if value == "":
                raise ValueError("Input cannot be empty.")
            
            return float(value)          # allow both int and float
            
        except ValueError as e:
            print(f"Invalid number: {e}. Please try again.")


def get_operator():
    """
    Keeps asking until a valid operator is entered.
    """
    valid_operators = ["+", "-", "*", "/", "%", "**"]
    
    while True:
        op = input("Enter operator (+, -, *, /, %, **): ").strip()
        
        if op in valid_operators:
            return op
        else:
            print("Invalid operator. Please enter one of: + - * / % **")


def calculate(num1, num2, operator):
    """
    Performs the actual calculation.
    Raises ZeroDivisionError if dividing/modulo by zero.
    """
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        return num1 / num2
    elif operator == "%":
        return num1 % num2
    elif operator == "**":
        return num1 ** num2


def main():
    print("=" * 45)
    print("         ROBUST CALCULATOR")
    print("=" * 45)

    try:
        # Step 1: Get inputs
        num1 = get_number("Enter first number: ")
        num2 = get_number("Enter second number: ")
        operator = get_operator()

        # Step 2: Perform calculation
        result = calculate(num1, num2, operator)

    except ZeroDivisionError:
        print("\nError: Cannot divide or modulo by zero.") # Handles division or modulo by zero
    
    except Exception as e:
        print(f"\nAn unexpected error occurred: {e}") # Handles any other unexpected error
    
    else:
        # This block runs only if no exception occurred
        print(f"\nResult: {num1} {operator} {num2} = {result}")
    
    finally:
        # This block always runs
        print("\nThank you for using the calculator.")


if __name__ == "__main__":
    main()
````

# Q4: CSV Data Validator

Create a CSV file containing both valid and invalid student records.
```
Input CSV
name,marks
Arsh,85
Rahul,abc
,90
Aman,105
Vijay,75
Neha,92
Ravi,-10
Priya,88
```
Your program should:

- Read every row.
- Validate the name.
- Validate the marks.
- Separate valid and invalid records.
- Record why invalid rows failed.
- Calculate statistics using only valid records.

Expected categories to detect
- Missing name
- Non-numeric marks
- Marks above 100
- Negative marks
- Valid record

Use exception handling where it makes sense rather than wrapping the entire program in one huge try block.

---

````python
import csv
import os


def read_csv(filename="students2.csv"):
    rows = []
    with open(filename, "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            rows.append(row)
    return rows


# Validate a single record
def validate_record(row):
    """
    Returns:
        (True, None)              → if valid
        (False, reason)           → if invalid
    """
    name = row.get("name", "").strip()
    marks_raw = row.get("marks", "").strip()

    # 1. Missing name
    if name == "":
        return False, "Missing name"

    # 2. Non-numeric marks
    try:
        marks = float(marks_raw)
    except ValueError:
        return False, "Non-numeric marks"

    # 3. Negative marks
    if marks < 0:
        return False, "Negative marks"

    # 4. Marks above 100
    if marks > 100:
        return False, "Marks above 100"

    # If all checks passed
    return True, None


# Separate valid and invalid records
def separate_records(rows):
    valid = []
    invalid = []

    for row in rows:
        is_valid, reason = validate_record(row)

        if is_valid:
            valid.append({
                "name": row["name"].strip(),
                "marks": float(row["marks"])
            })
        else:
            invalid.append({
                "name": row.get("name", "").strip(),
                "marks": row.get("marks", "").strip(),
                "reason": reason
            })

    return valid, invalid


# Calculate statistics (only on valid data)
def calculate_statistics(valid_records):
    if not valid_records:
        return {
            "total_valid": 0,
            "average": 0,
            "highest": 0,
            "lowest": 0
        }

    marks_list = [student["marks"] for student in valid_records]

    return {
        "total_valid": len(valid_records),
        "average": round(sum(marks_list) / len(marks_list), 2),
        "highest": max(marks_list),
        "lowest": min(marks_list)
    }


# Display Report
def generate_report(valid, invalid, stats):
    print("=" * 55)
    print("           CSV DATA VALIDATOR REPORT")
    print("=" * 55)

    print("\n--- Valid Records ---")
    if valid:
        for student in valid:
            print(f"Name: {student['name']:<10} | Marks: {student['marks']}")
    else:
        print("No valid records found.")

    print("\n--- Invalid Records ---")
    if invalid:
        for row in invalid:
            print(f"Name: '{row['name']}' | Marks: '{row['marks']}' → {row['reason']}")
    else:
        print("No invalid records found.")

    print("\n--- Statistics (Valid Records Only) ---")
    print(f"Total Valid Students : {stats['total_valid']}")
    print(f"Average Marks        : {stats['average']}")
    print(f"Highest Marks        : {stats['highest']}")
    print(f"Lowest Marks         : {stats['lowest']}")
    print("=" * 55)



if __name__ == "__main__":
    if not os.path.exists("students2.csv"):
        print("Error: students2.csv file not found.")
    else:
        rows = read_csv("students2.csv")
        valid_records, invalid_records = separate_records(rows)
        stats = calculate_statistics(valid_records)
        generate_report(valid_records, invalid_records, stats)
````
