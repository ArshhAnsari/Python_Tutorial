# Q5: Refactor a Function-Based Problem into a Class

Take your previous Inventory Manager and redesign it as:
```
Inventory
    ├── add_product()
    ├── remove_product()
    ├── update_quantity()
    ├── search_product()
    └── display_inventory()
```

Initial data
```python
inventory = {    "Laptop": 12,    "Mouse": 35,    "Keyboard": 20,    "Monitor": 8}
```

The exercise is not about the inventory logic — you've already done that.

The question is:

What changes when data and the functions operating on that data become one object?

---

````python
class Inventory:
    def __init__(self, products=None):
        self.products = products if products is not None else {}

    def add_product(self, name, quantity):
        if name in self.products:
            print(f"{name} already exists. Use update_quantity instead.")
        else:
            self.products[name] = quantity
            print(f"Added {name} with quantity {quantity}")

    def remove_product(self, name):
        if name in self.products:
            del self.products[name]
            print(f"Removed {name} from inventory")
        else:
            print(f"{name} not found in inventory")

    def update_quantity(self, name, quantity):
        if name in self.products:
            self.products[name] = quantity
            print(f"Updated {name} quantity to {quantity}")
        else:
            print(f"{name} not found in inventory")

    def search_product(self, name):
        if name in self.products:
            print(f"{name}: {self.products[name]} units")
        else:
            print(f"{name} not found in inventory")

    def display_inventory(self):
        if not self.products:
            print("Inventory is empty")
            return

        print("\n--- Inventory ---")
        for product, quantity in self.products.items():
            print(f"{product}: {quantity}")
        print("-----------------\n")


inventory = Inventory({
    "Laptop": 12,
    "Mouse": 35,
    "Keyboard": 20,
    "Monitor": 8
})


inventory.display_inventory()
inventory.add_product("Headphones", 15)
inventory.update_quantity("Mouse", 40)
inventory.search_product("Laptop")
inventory.remove_product("Monitor")
inventory.display_inventory()
````

# Q6: Encapsulated Bank Account

Create a BankAccount class where the balance cannot be freely modified from outside the class.

Provide controlled methods:
```
deposit()
withdraw()
get_balance()
```

Test
```
Initial balance = 10000

deposit(5000)
withdraw(3000)
withdraw(15000)
```

Try to directly modify the balance and observe what your design allows.

Then compare:
```
balance
_balance
__balance
```
and understand the Python conventions/name mangling involved.

---

````python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance        # public
        self._balance = balance       # protected (convention)
        self.__balance = balance      # private (name mangling)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print(f"Insufficient funds. Current Balance: {self.__balance}")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    def get_balance(self):
        return self.__balance



account = BankAccount(10000)

account.deposit(5000)     
account.withdraw(3000)    
account.withdraw(15000) 

print("Balance through method:", account.get_balance())


# ------------------------------------------------------------
# Comparing balance, _balance, __balance
# ------------------------------------------------------------
print("\n--- Access Comparison ---")

# 1. Public
print("Public balance:", account.balance)
account.balance = 999999
print("Modified public balance:", account.balance)

# 2. Protected (by convention only)
print("Protected _balance:", account._balance)
account._balance = 888888
print("Modified _balance:", account._balance)

# 3. Private (name mangling)
try:
    print("Private __balance:", account.__balance)
except AttributeError as e:
    print("Cannot access __balance directly:", e)

# Accessing private variable through name mangling
print("Private via name mangling:", account._BankAccount__balance)

print("\nActual balance (safe method):", account.get_balance())
````

Create a Person class with:
```

age
```

Use *@property* so that age cannot be assigned an invalid value.

Input
```
Valid:
22
35
18

Invalid:
-5
0
150
```
The class should reject invalid ages.

### Goal

Understand why *person.age = 25*, can still execute validation logic internally through a property.

---

````python
class Person:
    def __init__(self, name, age):
        self.name = name 
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")
        
    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, age):
        if 0 < age <=100:
            self._age = age
        else:
            print(f"Age: {age} | ValueError: Age must be between 1 and 100")

p = Person("John", 30)
p.display_info()
p.age = 35
print(p.age)
p.age = 150
p.display_info()
````

Create a Product class with:
```
name
price

```

Add validation so:
- price >= 0
- quantity >= 0

Input:
```python
name = "Keyboard"
price = 2500

```

Then test:
```python
price = -500
quantity = -10
```

This combines encapsulation with validation.

---

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, price):
        if price < 0:
            print(f"ValueError: Price cannot be negative!")
        else:
            self._price = price

    @property
    def quantity(self):
        return self._quantity

    @quantity.setter
    def quantity(self, quantity):
        if quantity < 0:
            print(f"ValueError: Quantity cannot be negative!")
        else:
            self._quantity = quantity

    def display_info(self):
        print(f"Product Name: {self.name}, Price: {self.price}, Quantity: {self.quantity}")

# ---------- #
k = Product("keyboard", 2500, 20)
k.display_info()

k.price = -500
k.quantity = -10
k.display_info()
````

Input

# Q13: Money Class [Operator Overloading & Dunder Methods]

Create:
```text
Money
```

with:
```text
amount
currency
```

Implement:
```python
__add__()
__eq__()
__str__()
__repr__()
```

Input
```python
m1 = Money(1000, "INR")
m2 = Money(500, "INR")
m3 = Money(1000, "INR")
```

Test:
```python
m1 + m2
m1 == m3
str(m1)
repr(m1)
```

Goal
Understand the connection:

```
m1 + m2
    ↓
m1.__add__(m2)

m1 == m3
    ↓
m1.__eq__(m3)
```

---

````python
class Money:
    def __init__(self, amount, currency):
        self.amount = amount
        self.currency = currency

    def __add__(self, other):
        if self.currency != other.currency:
            raise ValueError("Currencies must be the same to add")
        return Money(self.amount + other.amount, self.currency)

    def __eq__(self, other):
        return self.amount == other.amount and self.currency == other.currency

    def __str__(self):
        return f"{self.amount} {self.currency}"

    def __repr__(self):
        return f"Money(amount={self.amount}, currency='{self.currency}')"


m1 = Money(1000, "INR")
m2 = Money(500, "INR")
m3 = Money(1000, "INR")

print(m1 + m2)      
print(m1 == m3)    
print(str(m1))      
print(repr(m1))     
````

Input:
```

```

Try creating an object directly from PaymentMethod.
Understand why the abstract class cannot be used as a normal concrete implementation.

---

````python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")


class BankTransfer(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} using Bank Transfer")


CreditCard().pay(2500)
UPI().pay(1500)
BankTransfer().pay(5000)


# Trying to create object from abstract class
try:
    payment = PaymentMethod()
except TypeError as e:
    print("\nError:", e)
````

# Q16: Instance vs Class vs Static Method

Create a Employee class containing:
```
name
employee_id
department
```

Implement:
```
instance method
classmethod
staticmethod
```

For example:
```python
employee.display_details()
Employee.from_string("101,Arsh,Backend")
Employee.is_valid_department("Backend")
```

Input
```
"101,Arsh,Backend"
"102,Rahul,Frontend"
"103,Neha,Data"
```

---

````python
class Employee:
    valid_departments = ["Backend", "Frontend", "Data", "HR", "Finance"]

    def __init__(self, employee_id, name, department):
        self.employee_id = employee_id
        self.name = name
        self.department = department

    # ---------- Instance Method ----------
    def display_details(self):
        print(f"ID: {self.employee_id}, Name: {self.name}, Department: {self.department}")

    # ---------- Class Method ----------
    @classmethod
    def from_string(cls, data):
        employee_id, name, department = data.split(",")
        return cls(employee_id, name, department)

    # ---------- Static Method ----------
    @staticmethod
    def is_valid_department(department):
        return department in Employee.valid_departments


# Input
emp1 = Employee.from_string("101,Arsh,Backend")
emp2 = Employee.from_string("102,Rahul,Frontend")
emp3 = Employee.from_string("103,Neha,Data")

emp1.display_details()
emp2.display_details()
emp3.display_details()

print("Is Backend valid?", Employee.is_valid_department("Backend"))
print("Is Sales valid?", Employee.is_valid_department("Sales"))
````

# Q17: Design a Delivery System

Create a system with:
```
User
   ↓
Customer

Delivery
   ↓
BikeDelivery
TruckDelivery
```

Use:
```
classes
objects
__init__
instance variables
class variables
inheritance
method overriding
polymorphism
encapsulation
@property
abstract base class

dunder methods
```

Input:
```

ID = 101
Name = Arsh
```
```
Delivery 1:
Type = Bike
Distance = 8 km
```
```
Delivery 2:
Type = Truck
Distance = 35 km
```

The system should calculate delivery cost differently for each delivery type.

---

````python
from abc import ABC, abstractmethod


# =========================
# USER / CUSTOMER
# =========================
class User:
    def __init__(self, id, name):
        self.id = id
        self.name = name

    def __str__(self):
        return f"User(ID={self.id}, Name={self.name})"


class Customer(User):
    def __init__(self, id, name):
        super().__init__(id, name)

    def __str__(self):
        return f"Customer(ID={self.id}, Name={self.name})"


# =========================
# DELIVERY SYSTEM
# =========================
class Delivery(ABC):
    company_name = "FastDeliver"     # class variable
    total_deliveries = 0             # class variable

    def __init__(self, distance):
        if not Delivery.is_valid_distance(distance):
            raise ValueError("Distance must be positive")
        self._distance = distance    # encapsulation
        Delivery.total_deliveries += 1

    @property
    def distance(self):              # property
        return self._distance

    @abstractmethod
    def calculate_cost(self):
        pass

    @staticmethod
    def is_valid_distance(distance): # static method
        return distance > 0

    @classmethod
    def get_total_deliveries(cls):   # class method
        return cls.total_deliveries

    def __str__(self):
        return f"{self.__class__.__name__}(distance={self.distance} km)"


class BikeDelivery(Delivery):
    rate_per_km = 10

    def calculate_cost(self):        # method overriding
        cost = self.distance * self.rate_per_km
        print(f"Bike Delivery: {self.distance} km × ₹{self.rate_per_km} = ₹{cost}")
        return cost


class TruckDelivery(Delivery):
    rate_per_km = 20

    def calculate_cost(self):        # method overriding
        cost = self.distance * self.rate_per_km
        print(f"Truck Delivery: {self.distance} km × ₹{self.rate_per_km} = ₹{cost}")
        return cost


# =========================
# DEMO / INPUT
# =========================
customer = Customer(101, "Arsh")
print(customer)

# Delivery objects
bike = BikeDelivery(8)
truck = TruckDelivery(35)

# Polymorphism
payments = [bike, truck]
print("\n--- Calculating Costs ---")
for delivery in payments:
    delivery.calculate_cost()

print("\nTotal deliveries:", Delivery.get_total_deliveries())
print("Company:", Delivery.company_name)

# Abstract class cannot be instantiated
try:
    d = Delivery(10)
except TypeError as e:
    print("\nAbstract class error:", e)
````
