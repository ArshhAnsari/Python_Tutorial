````python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display_info(self):
        print(f"Student Name: {self.name}, Age: {self.age}, Course: {self.course}")

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age}, course='{self.course}')"

students_data = [
    ("Arsh", 22, "Python"),
    ("Rahul", 23, "Django"),
    ("Neha", 21, "DSA")
]

# Create and store objects
students = []
for name, age, course in students_data:
    students.append(Student(name, age, course))

# Display all
for student in students:
    student.display_info()


# The List stores the references to the Student Class objects, not the objects themselves. 
# So, when we print the List, it will show the references to the objects, not the actual objects.
# print(students) uses the default object representation
# It does not automatically show name, age, course
print(students) 
````

---

What you print | What you see|
---------- | ----------|

With __repr__ | Readable student details |

````python
print(students) # This will use the __repr__ method to display the List of student objects in readable format.
````

# Q3: Object-Based Counter

Create a Counter class with:
```python
increment()
decrement()
reset()
display()
```

Input
```python
Start_value = 10
```

Operations:
```
increment
increment
decrement
increment
reset
increment
```

---

````python
class Counter:
    def __init__(self, start_value=0):
        self.value = start_value

    def increment(self):
        self.value += 1

    def decrement(self):
        self.value -= 1

    def reset(self):
        self.value = 0

    def display(self):
        print(f"Current Counter Value: {self.value}")


counter = Counter(10)

counter.increment()   # 11
counter.increment()   # 12
counter.decrement()   # 11
counter.increment()   # 12
counter.reset()       # 0
counter.increment()   # 1

counter.display()
````

- Deposit 5000

The last withdrawal should fail because of insufficient balance.

---

````python
class BankAccount:
    def __init__(self, account_number, holder_name, balance = 0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: {amount}, New Balance: {self.balance}")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn: {amount}, New Balance: {self.balance}")

    def display_balance(self):
        print(f"Account Number: {self.account_number}, Holder Name: {self.holder_name} , Balance: {self.balance}" )


t = BankAccount(1001, "Arsh", 10000)
u = BankAccount(1002, "Rahul", 15000)
v = BankAccount(1003, "Neha", 20000)


t.display_balance()
t.deposit(5000)
t.withdraw(2500)
t.deposit(3000)
t.withdraw(12000)

t.display_balance()
````

# Q5: Refactor a Function-Based Problem into a Class

Take your previous Inventory Manager and redesign it as:
```
Inventory
    ├── add_product()
    ├── remove_product()
    ├── update_quantity()
    ├── search_product()
    └── display_inventory()
```

Initial data
```python
inventory = {    "Laptop": 12,    "Mouse": 35,    "Keyboard": 20,    "Monitor": 8}
```

The exercise is not about the inventory logic — you've already done that.

The question is:

What changes when data and the functions operating on that data become one object?

---

````python
class Inventory:
    def __init__(self, products=None):
        self.products = products if products is not None else {}

    def add_product(self, name, quantity):
        if name in self.products:
            print(f"{name} already exists. Use update_quantity instead.")
        else:
            self.products[name] = quantity
            print(f"Added {name} with quantity {quantity}")

    def remove_product(self, name):
        if name in self.products:
            del self.products[name]
            print(f"Removed {name} from inventory")
        else:
            print(f"{name} not found in inventory")

    def update_quantity(self, name, quantity):
        if name in self.products:
            self.products[name] = quantity
            print(f"Updated {name} quantity to {quantity}")
        else:
            print(f"{name} not found in inventory")

    def search_product(self, name):
        if name in self.products:
            print(f"{name}: {self.products[name]} units")
        else:
            print(f"{name} not found in inventory")

    def display_inventory(self):
        if not self.products:
            print("Inventory is empty")
            return

        print("\n--- Inventory ---")
        for product, quantity in self.products.items():
            print(f"{product}: {quantity}")
        print("-----------------\n")


inventory = Inventory({
    "Laptop": 12,
    "Mouse": 35,
    "Keyboard": 20,
    "Monitor": 8
})


inventory.display_inventory()
inventory.add_product("Headphones", 15)
inventory.update_quantity("Mouse", 40)
inventory.search_product("Laptop")
inventory.remove_product("Monitor")
inventory.display_inventory()
````

# Q6: Encapsulated Bank Account

Create a BankAccount class where the balance cannot be freely modified from outside the class.

Provide controlled methods:
```
deposit()
withdraw()
get_balance()
```

Test
```
Initial balance = 10000

deposit(5000)
withdraw(3000)
withdraw(15000)
```

Try to directly modify the balance and observe what your design allows.

Then compare:
```
balance
_balance
__balance
```
and understand the Python conventions/name mangling involved.

---

````python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance        # public
        self._balance = balance       # protected (convention)
        self.__balance = balance      # private (name mangling)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print(f"Insufficient funds. Current Balance: {self.__balance}")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    def get_balance(self):
        return self.__balance



account = BankAccount(10000)

account.deposit(5000)     
account.withdraw(3000)    
account.withdraw(15000) 

print("Balance through method:", account.get_balance())


# ------------------------------------------------------------
# Comparing balance, _balance, __balance
# ------------------------------------------------------------
print("\n--- Access Comparison ---")

# 1. Public
print("Public balance:", account.balance)
account.balance = 999999
print("Modified public balance:", account.balance)

# 2. Protected (by convention only)
print("Protected _balance:", account._balance)
account._balance = 888888
print("Modified _balance:", account._balance)

# 3. Private (name mangling)
try:
    print("Private __balance:", account.__balance)
except AttributeError as e:
    print("Cannot access __balance directly:", e)

# Accessing private variable through name mangling
print("Private via name mangling:", account._BankAccount__balance)

print("\nActual balance (safe method):", account.get_balance())
````

# Q7: Property-Based Validation

Create a Person class with:
```
name
age
```

Use *@property* so that age cannot be assigned an invalid value.

Input
```
Valid:
22
35
18

Invalid:
-5
0
150
```
The class should reject invalid ages.

### Goal

Understand why *person.age = 25*, can still execute validation logic internally through a property.

---

````python
class Person:
    def __init__(self, name, age):
        self.name = name 
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")
        
    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, age):
        if 0 < age <=100:
            self._age = age
        else:
            print(f"Age: {age} | ValueError: Age must be between 1 and 100")

p = Person("John", 30)
p.display_info()
p.age = 35
print(p.age)
p.age = 150
p.display_info()
````

# Q8: Encapsulated Product

Create a Product class with:
```
name
price
quantity
```

Add validation so:
- price >= 0
- quantity >= 0

Input:
```python
name = "Keyboard"
price = 2500
quantity = 20
```

Then test:
```python
price = -500
quantity = -10
```

This combines encapsulation with validation.

---

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, price):
        if price < 0:
            print(f"ValueError: Price cannot be negative!")
        else:
            self._price = price

    @property
    def quantity(self):
        return self._quantity

    @quantity.setter
    def quantity(self, quantity):
        if quantity < 0:
            print(f"ValueError: Quantity cannot be negative!")
        else:
            self._quantity = quantity

    def display_info(self):
        print(f"Product Name: {self.name}, Price: {self.price}, Quantity: {self.quantity}")

# ---------- #
k = Product("keyboard", 2500, 20)
k.display_info()

k.price = -500
k.quantity = -10
k.display_info()
````

# Q9: Employee Hierarchy

Create:
```
Employee
    ↓
Developer
    ↓
Manager
```

The base class should contain common information:
```
name
employee_id
salary
```

Developer should add:
```
programming_language
```

Manager should add:
```
team_size
```

Input:

Developer:
```
    Name = Arsh
    ID = 101
    Salary = 70000
    Language = Python
```
Manager:
```
    Name = Rahul
    ID = 102
    Salary = 90000
    Team Size = 8
```

Understand what belongs in the parent and what belongs in the child.

````python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"ID: {self.employee_id}")
        print(f"Salary: {self.salary}")


class Developer(Employee):
    def __init__(self, name, employee_id, salary, programming_language):
        super().__init__(name, employee_id, salary)
        self.programming_language = programming_language

    def display_info(self):
        super().display_info()
        print(f"Language: {self.programming_language}")


class Manager(Employee):
    def __init__(self, name, employee_id, salary, team_size):
        super().__init__(name, employee_id, salary)
        self.team_size = team_size

    def display_info(self):
        super().display_info()
        print(f"Team Size: {self.team_size}")


# Test Input
dev = Developer("Arsh", 101, 70000, "Python")
mgr = Manager("Rahul", 102, 90000, 8)

print("--- Developer ---")
dev.display_info()

print("\n--- Manager ---")
mgr.display_info()
````

# Q10: Multi-Level Inheritance
Create:
```
Vehicle
   ↓
  Car
   ↓
ElectricCar
```

Input
```
Brand = Tesla
Model = Model 3
Battery Capacity = 60
```

Give each level at least one responsibility.

---

````python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def show_brand(self):
        print(f"Brand: {self.brand}")


class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def show_model(self):
        print(f"Model: {self.model}")


class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

    def show_battery(self):
        print(f"Battery Capacity: {self.battery_capacity} kWh")

    def display_info(self):
        self.show_brand()
        self.show_model()
        self.show_battery()


car = ElectricCar("Tesla", "Model 3", 60)
car.display_info()
````
