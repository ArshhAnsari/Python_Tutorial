# Q1: Refactor the Student Data Analyzer

You previously created a Student Data Analyzer using lists and dictionaries.

Refactor that program so that each responsibility is handled by a separate function.

Create functions for:
```python
calculate_average()
find_highest_scorer()
find_lowest_scorer()
find_top_students()
find_passing_students()
calculate_overall_average()
```
Then create a main() function that calls them and produces the final report.

Input
```python
students = [
    {"name": "Arsh", "marks": [85, 78, 92]},
    {"name": "Rahul", "marks": [72, 88, 75]},
    {"name": "Aman", "marks": [95, 91, 89]},
    {"name": "Vijay", "marks": [60, 68, 64]},
    {"name": "Neha", "marks": [82, 79, 85]}
]
```

passing_average = 70

top_n = 3

Your functions should work with this data without putting the entire logic inside main().

---

````python
def calculate_average(students):
    for s in students:
        avg = round((sum(s['marks'])/len(s['marks'])), 2)
        print(f"  • {s['name']:<8} : {avg:.2f}")

def find_highest_scorer(students):
    """Finds and prints the student with the highest total score."""
    highscorer = max(students, key=lambda s: sum(s['marks']))
    total_marks = sum(highscorer['marks'])
    print(f"  🏆 {highscorer['name']} with a total score of {total_marks} {highscorer['marks']}")

def find_lowest_scorer(students):
    """Finds and prints the student with the lowest total score."""
    lowscorer = min(students, key=lambda s: sum(s['marks']))
    total_marks = sum(lowscorer['marks'])
    print(f"  🔻 {lowscorer['name']} with a total score of {total_marks} {lowscorer['marks']}")    

def find_top_students(students,n):
    topstd = sorted(students, key=lambda s: sum(s['marks']), reverse=True)
    for i in range(n):
        print(f"The {i+1}'s position secured by {topstd[i]['name']:<8} | Total score {sum(topstd[i]['marks'])}")

def find_passing_students(students, n):
    for i in students:
        avg = sum(i['marks'])/len(i['marks'])
        if avg > n:
            print(f"  ✓ {i['name']:<8} : Passed (Average: {avg:.2f})")
    

def calculate_overall_average(students):
    oversum = 0
    overlen = 0
    for i in students:
        oversum += sum(i['marks'])
        overlen += len(i['marks'])

    return oversum/overlen


if __name__ == "__main__":

    students = [ {"name": "Arsh", "marks": [85, 78, 92]}, 
                {"name": "Rahul", "marks": [72, 88, 75]}, 
                {"name": "Aman", "marks": [95, 91, 89]}, 
                {"name": "Vijay", "marks": [60, 68, 64]}, 
                {"name": "Neha", "marks": [82, 79, 85]}
    ]

    print("==========================================")
    print("        STUDENT PERFORMANCE REPORT        ")
    print("==========================================")

    print("\n--- Student Averages ---")
    calculate_average(students)

    print("\n--- Highest Scorer ---")
    find_highest_scorer(students)

    print("\n--- Lowest Scorer ---")
    find_lowest_scorer(students)

    top_n = 3
    print(f"\n--- Top {top_n} Students ---")
    find_top_students(students, top_n)

    passing_average = 70
    print(f"\n--- Passing Students (Average > {passing_average}) ---")
    find_passing_students(students, passing_average)

    print("\n📊 The Overall Class Average is: ",end="")
    print(calculate_overall_average(students))
````

# Q2: HCF & LCM Utility

Create reusable functions:

- gcd(a, b)
- lcm(a, b)

Then extend the program so it can calculate HCF and LCM for three or more numbers.

Finally create:
- calculate_gcd(*numbers)
- calculate_lcm(*numbers)

Input 1
> a = 12
> b = 18

Input 2
> numbers = [12, 18, 24]

Input 3
> numbers = [15, 25, 35, 45]

Your final version should be able to work with any number of values passed through *args.

---

### Explanation:

GCD (Greatest Common Divisor): It is the largest positive integer that divides two or more numbers evenly without leaving a remainder. Also called HCF (Highest Common Factor). 
For 12 and 18, the common divisors are 1, 2, 3, and 6. The GCD is 6.

> The most efficient way to compute the GCD of two numbers is the Euclidean Algorithm.
- The Logic: The GCD of two numbers $a$ and $b$ is the same as the GCD of $b$ and the remainder of $a$ divided by $b$ ($a \pmod b$). You repeat this replacement until the remainder becomes $0$. The last non-zero remainder is your GCD.
- Trace Example (gcd(48, 18)):
1. $48 \pmod{18} = 12 \implies \text{gcd}(18, 12)$
2. $18 \pmod{12} = 6 \implies \text{gcd}(12, 6)$
3. $12 \pmod 6 = 0 \implies \text{gcd}(6, 0)$
4. Base case reached ($b = 0$): return 6.

---

LCM (Least Common Multiple): LCM is the smallest positive integer that is divisible by each of those numbers. 
For 4 and 6, the multiples of 4 are (4, 8, 12...) and 6 are (6, 12, 18...). The smallest overlap is 12.

> Instead of searching multiples manually, use the mathematical relationship between GCD and LCM:
$$\text{lcm}(a, b) = \frac{\vert{}a \times b\vert{}}{\gcd(a, b)}$$

- Since you already built gcd(a, b), calculating LCM is just multiplying the two numbers and dividing by their GCD.
---

````python
def gcd(a,b):
    while b!=0:
        a,b = b, (a % b)
    return a

def lcm(a,b):
    numerator = a * b
    denominator = gcd(a,b)
    return numerator // denominator

if __name__ == "__main__":
    num1 = int(input("Enter the number: "))
    num2 = int(input("Enter the number: "))
    task = input("What do you want to calculate: 1.GCD | 2.LCM | 3.Both")

    match task:
        case "1":
            print(f"The GCD for the values {num1} & {num2} is: {gcd(num1,num2)}")
        case "2":
            print(f"The LCM for the values {num1} & {num2} is: {lcm(num1,num2)}")
        case "3":
            print(f"The GCD & LCM value for {num1} & {num2} is: GCD:{gcd(num1,num2)} , LCM:{lcm(num1,num2)}")
        case _:
            print("Invalid task choice!")


    
````

---

GCD and LCM operations are associative. This means:

$$\gcd(x, y, z) = \gcd(\gcd(x, y), z)$$

$$\text{lcm}(x, y, z) = \text{lcm}(\text{lcm}(x, y), z)$$

---


````python
# --- Helpers Functions ---
def gcd_two(a, b):
    '''
    Repeatedly finding the remainder using modulo (a % b) until it zeroes out. 
    Whatever is left over at the end is the Greatest Common Divisor.
    '''
    while b != 0:
        a, b = b, (a % b)
    return a

def lcm_two(a, b):
    '''
    Multiplies them together, and divides them by their GCD 
    to instantly gives the Least Common Multiple.
    '''
    return (a * b) // gcd_two(a, b)


# --- For N numbers ---

def calculate_gcd(*numbers):
    if not numbers:
        return 0
    
    result = numbers[0] # Very first number and call it our result (our running total).
    for x in numbers[1:]: # Looping over to the rest of the numbers
        result = gcd_two(result, x) # Passing to the helper function, the answer updates itself on the fly. 
                                    # By the time loop finishes checking the last number, 
                                    # result holds the final answer for the entire group.
    return result

def calculate_lcm(*numbers):
    if not numbers:
        return 0
        
    result = numbers[0]
    for x in numbers[1:]:
        result = lcm_two(result, x)
    return result


if __name__ == "__main__":
    numbers = [int(x) for x in input("Enter numbers separated by spaces (e.g., 12 18 24): ").split()]
    
    task = input("What do you want to calculate: 1.GCD | 2.LCM | 3.Both: ")

    match task:
        case "1":
            print(f"The GCD values for the number's - {numbers}: {calculate_gcd(*numbers)}")
        case "2":
            print(f"The LCM values for the number's - {numbers}: {calculate_lcm(*numbers)}")
        case "3":
            print(f"For the number's {numbers} the values are: \nThe GCD is: {calculate_gcd(*numbers)} , LCM is: {calculate_lcm(*numbers)}")
        case _:
            print("Invalid task choice!")
````
