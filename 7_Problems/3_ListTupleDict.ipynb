# Q1: Student Marks Analyzer

Given:

students = [
    {"name": "A", "marks": 85},
    {"name": "B", "marks": 72},
    {"name": "C", "marks": 91},
    {"name": "D", "marks": 64}
]

Find:

- Highest scorer.
- Lowest scorer.
- Average marks.
- Top 3 students.
- Students who scored above the average.

---

````python

students = [
    {"name": "A", "marks": 85},
    {"name": "B", "marks": 72},
    {"name": "C", "marks": 91},
    {"name": "D", "marks": 64}
]

# highest_marks = max(student["marks"] for student in students)
highest_marks = [{student["marks"] : student["name"]} for student in students if student["marks"] == max(student["marks"] for student in students)]

# lowest_marks = min(student["marks"] for student in students)
lowest_marks = [{student["marks"] : student["name"]} for student in students if student["marks"] == min(student["marks"] for student in students)]

student_sorted =  sorted([student["marks"] for student in students], reverse=True)
top_3 = student_sorted[:3]
top_3_students = [
    {student["name"]: student["marks"]} 
    for student in students 
    if student["marks"] in top_3
]

avg_marks= sum(student["marks"] for student in students)/len(students)
above_avg_students = [
    {student["name"]: student["marks"]} 
    for student in students 
    if student["marks"] > avg_marks
]


print(f"Highest marks scorer: {highest_marks}")
print(f"Lowest marks scorer: {lowest_marks}")
print(f"Average marks: {avg_marks:.2f}")

print("\nTop 3 scorers:")
print(top_3_students)
print("\nStudents who scored above average:")
print(above_avg_students)
````

````python
# Prerequisite: Understand lambda functions before using them to sort students by "marks" in descending order
sorted_students = sorted(students, key=lambda x: x["marks"], reverse=True)
print(sorted_students)
````

# Q2: List Statistics

Do not use min(), max(), or sort().

---

Given a list containing duplicates, create a new list containing only unique elements.

Solve it without using set().

Then solve it again using a set.

---

````python
lst = [1,2,4,2,5,1,3,2,6,8]

deduplicate_lst = []

for i in lst:
    if i not in deduplicate_lst:
        deduplicate_lst.append(i)

print(deduplicate_lst)

print(set(lst))
````

# Q4: Tuple Data Processing

Find:

- Highest scorer.
- Lowest scorer.
- Average marks.
- Students scoring above 80.

---

````python
students = [
    ("Arjun", 85),
    ("Rahul", 72),
    ("Aman", 91),
    ("Vijay", 64)
]

lst=[]
for i in students:
    lst.append({ 'name': i[0],'marks': i[1] } )
    # lst.append({**{i[0]: i[1]}})

# lst = [{'name': name, 'marks': marks} for name, marks in students]

# Calculate max and min only once
max_marks = max(student['marks'] for student in lst)
min_marks = min(student['marks'] for student in lst)

highest_students = [student for student in lst if student['marks'] == max_marks]
lowest_students  = [student for student in lst if student['marks'] == min_marks]

avg_marks= sum(i['marks'] for i in lst)/len(lst)
scoring= [i for i in lst if i['marks'] > 80]

print("All students:", lst)
print("Highest scorer(s):", highest_students)
print("Lowest scorer(s):", lowest_students)
print("Average marks:", round(avg_marks))
print("Students scoring above 80:", scoring)
````

### Step-by-step breakdown for ({**{i[0]: i[1]} :


Assume i = ("Arjun", 85)

1. Inner part:
```
{i[0]: i[1]}
```
→ Creates a dictionary:
{"Arjun": 85}

2. Unpacking with **:
```
**{"Arjun": 85}
```
→ This unpacks the dictionary into key-value pairs:
Arjun = 85

3. Outer curly braces { }:

```
{ Arjun = 85 }
```

→ When you write key-value pairs inside {}, Python automatically packs them back into a dictionary.

```
{**{"Arjun": 85}}     is same as     {"Arjun": 85}
```

# Q5: Character Frequency Analyzer

Given a string, create a dictionary containing the frequency of every character.

Example:

Input:  programming

Output: {'p': 1, 'r': 2, ...}

Then identify:

- Most frequent character.
- Least frequent character.
- Characters appearing exactly once.

---

````python
str = "programming"
dict_str={}

for i in str:
    count=0
    for j in str:
        if i == j:
            count+=1 
    dict_str[i] = count
    

print(dict_str)

max_char = max(dict_str.values())
min_char = min(dict_str.values())

most_freq = [char for char, freq in dict_str.items() if freq == max_char]
least_freq = [char for char, freq in dict_str.items() if freq == min_char]
exact_ones = [char for char, freq in dict_str.items() if freq == 1]

print("The character appearing with most frequency:",most_freq)
print("The character appearing with least frequency:",least_freq)
print("Character appearing exactly once:",exact_ones)
# Using max() with key , but only gives the first max freq char
# most_freq_char = [max(dict_str, key=dict_str.get)]
# print("Most frequent character:", most_freq_char)

````

### How it works step by step:

Suppose we have this dictionary:

```python
dict_str = {
    'p': 1,
    'r': 2,
    'o': 1,
    'g': 2,
    'a': 1,
    'm': 2,
    'i': 1,
    'n': 1
}
```

1. What does max(dict_str) normally do?

max(dict_str)

→ It compares the keys alphabetically and returns the biggest key ('r').But we don’t want that. We want the key whose value is maximum.

2. Role of key=dict_str.get

The key parameter tells max():
“Instead of comparing the items directly, compare the result of this function.”

dict_str.get is a function that takes a key and returns its value.

---

So behind the scenes, Python does something like this:

| Character | dict_str.get(character) | Value used for comparison |
|-----------|--------------------------|---------------------------|
| 'p'       | dict_str.get('p')        | 1                         |
| 'r'       | dict_str.get('r')        | 2                         |
| 'o'       | dict_str.get('o')        | 1                         |
| 'g'       | dict_str.get('g')        | 2                         |
| 'a'       | dict_str.get('a')        | 1                         |
| 'm'       | dict_str.get('m')        | 2                         |
| 'i'       | dict_str.get('i')        | 1                         |
| 'n'       | dict_str.get('n')        | 1                         |


Now max() finds the maximum value (which is 2) and returns the corresponding key.

---

Result:
```python
most_freq_char = 'r'   # (or 'g' or 'm' depending on which one appears first)
```
> Note: If multiple characters have the same highest frequency, max() returns the first one it finds.

# Q6: Word Frequency Counter

Given a sentence, create a dictionary containing the frequency of every word.

Then find the most frequently used word.

---

````python
sentence = "Python is easy and Python is powerful, Python is also very popular!"

sentence_dict = ""

for char in sentence:
    if char.isalpha() or char.isspace():
        sentence_dict += char.lower()

words = sentence_dict.split()

word_dict = {}
for word in words:
    if word in word_dict:
        word_dict[word] += 1
    else:
        word_dict[word] = 1

print(f"The given string of sentence: '{sentence_dict}'")

print(f"The frequency of each word in the sentence are: {word_dict}")
````

# Q8: Inventory Manager

Create an inventory represented using a dictionary.

The program should allow:

- Add product.
- Update quantity.
- Remove product.
- Search product.
- Display all products.
- Find the product with the highest quantity.

---

````python
inventory = {}

while True:
    print("\n========== Inventory Manager ==========")
    print("1. Add product")
    print("2. Update quantity")
    print("3. Remove product")
    print("4. Search product")
    print("5. Display all products")
    print("6. Find product with highest quantity")
    print("7. Exit")
    
    option = int(input("Choose an option (1-7): "))

    if option == 1:
        product = input("Enter the product name: ").strip()
        if product in inventory:
            print("Product already exists!")
        else:
            inventory[product] = 0
            print(f"'{product}' added successfully!")

    elif option == 2:
        product = input("Enter the product name: ").strip()
        if product in inventory:
            quantity = int(input(f"Enter quantity to add for {product}: "))
            inventory[product] += quantity
            print(f"Updated! {product} now has {inventory[product]} units.")
        else:
            print("Product not found! Please add it first.")

    elif option == 3:
        product = input("Enter the product name to remove: ").strip()
        if product in inventory:
            inventory.pop(product)
            print(f"'{product}' removed successfully!")
        else:
            print("Product not found!")

    elif option == 4:
        product = input("Enter the product name to search: ").strip()
        if product in inventory:
            print(f"Product: {product} | Quantity: {inventory[product]}")
        else:
            print(f"Product '{product}' not found!")

    elif option == 5:
        if inventory:
            print("\nCurrent Inventory:")
            for product, qty in inventory.items():
                print(f"  → {product}: {qty}")
        else:
            print("Inventory is empty!")

    elif option == 6:
        if inventory:
            max_qty = max(inventory.values())
            highest = [name for name, qty in inventory.items() if qty == max_qty]
            print(f"Highest quantity: {max_qty}")
            print(f"Product(s): {highest}")
        else:
            print("Inventory is empty!")

    elif option == 7:
        print("Exiting Inventory Manager. Goodbye!")
        break

    else:
        print("Invalid option! Please choose between 1-7.")
````
