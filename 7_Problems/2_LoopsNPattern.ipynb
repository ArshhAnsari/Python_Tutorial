# Q1: Reverse & Palindrome

Given an integer:

- Reverse its digits.
- Check whether the original number is a palindrome.

---

```python
og_number = int(input("Enter the number:"))
print(f"The original number is: {og_number}")
number = og_number
rev = 0

while number > 0:
    digit = number % 10
    rev = rev * 10 + digit
    number = number // 10

print(f"The reverse of the number is: {rev}")

if rev == og_number:
    print("The number is a palindrome")
else:
    print("The number is not a palindrome")
```

# Q2: Fibonacci Generator

Generate the first n Fibonacci numbers.

Then modify the program to:

- Print only Fibonacci numbers below a given limit.
- Calculate their sum.

---

```python
n=int(input("Enter the number of terms:"))

first=0
second=1
lst=[]

for i in range(n+1):
    lst.append(first)

    first, second = second, first+second

print(f"The fibonacci series is: {lst}")
print(f"Fibonacci number at index {n} is: {lst[n]}")

```

```python
limit = int(input("Enter the limit: "))

first = 0
second = 1
total = 0
lst = []

# Loop strictly while numbers are below the limit
while first < limit:
    lst.append(first)
    total += first
    first, second = second, first + second

# Display results
print("Fibonacci numbers below the limit:", *lst)
print(f"Sum: {total}")
```

# Q4: Mini Challenge: Guessing Game

Generate a secret number and allow the user to repeatedly guess it.

The program should:

- Tell the user whether the guess is too high or too low.
- Count attempts.
- Stop when the user guesses correctly.
- Allow the user to quit.

---

```python
import random;

randn = random.randint(1, 100)

print("Welcome to the Number Guessing Game!")

count = 0
while(True):
    guess = int(input("Guess the number between 1 and 100: "))

    if guess < randn: 
        print("Too low! Try again.")
        count += 1
    elif guess > randn: 
        print("Too high! Try again.")
        count += 1
    else:
        print("Congratulations! You guessed the number.")
        print(f"It took you {count} attempts to guess the number {randn}.")
        print(f"Do you want to play again? (yes/no)")
        if input().lower() == "yes":
            randn = random.randint(1, 100)
            count = 0
            continue
        else:
            print("Thank you for playing!")
            break

```
